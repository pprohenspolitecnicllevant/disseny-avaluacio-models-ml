## El cas: Ca Nostra Lloguers

**Ca Nostra Lloguers** és una immobiliària de Manacor que gestiona pisos i cases de
lloguer a tota la comarca de Llevant. Quan un propietari els porta un habitatge nou,
l'agent li proposa un preu de lloguer mensual mirant-se'l a ull, i cada agent ho fa
a la seva manera.

La gerent vol un model que **proposi el lloguer mensual d'un habitatge** a partir de
les seves característiques, i t'ha passat els 1.600 anuncis que es van publicar el
2025 en un portal de la comarca.

> ⚠️ **En aquest fitxer hi ha dues trampes.** Hi ha **una parella de columnes que
> diuen el mateix**, i hi ha **una columna que té correlació amb el lloguer però que
> no en pot ser cap causa**. Trobar-les és part dels exercicis. Per a la segona,
> fes-te aquesta pregunta: **si canvio aquest valor, canviaria de veritat el preu
> del pis?**

### Les columnes

| Columna | Què és |
|---|---|
| `id_anunci` | número de l'anunci al portal |
| `localitat` | on és l'habitatge |
| `tipus` | pis, casa, àtic o estudi |
| `estat` | a reformar, bon estat, reformat o obra nova |
| `metres_quadrats` | superfície útil, en m² |
| `peus_quadrats` | superfície que diu l'anunciant, en peus quadrats |
| `habitacions` | nombre de dormitoris (els estudis en tenen 0) |
| `banys` | nombre de banys |
| `planta` | planta on és l'habitatge (0 = planta baixa; les cases sempre 0) |
| `ascensor` | 1 si l'edifici té ascensor, 0 si no |
| `terrassa` | 1 si té terrassa o jardí, 0 si no |
| `moblat` | 1 si es lloga moblat, 0 si no |
| `distancia_mar_km` | distància en línia recta fins a la platja més propera, en km |
| `any_construccio` | any en què es va construir l'edifici |
| `lloguer_mensual` | **lloguer mensual de l'anunci, en euros**: el que volem predir |
