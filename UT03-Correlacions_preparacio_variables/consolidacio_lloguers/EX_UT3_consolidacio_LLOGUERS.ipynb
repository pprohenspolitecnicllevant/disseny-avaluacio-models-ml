## El cas: Ca Nostra Lloguers

**Ca Nostra Lloguers** és una immobiliària de Manacor que gestiona pisos i cases de
lloguer a tota la comarca de Llevant. Quan un propietari els porta un habitatge nou,
l'agent li proposa un preu de lloguer mensual mirant-se'l a ull, i cada agent ho fa
a la seva manera.

La gerent vol un model que **proposi el lloguer mensual d'un habitatge** a partir de
les seves característiques, i t'ha passat els 1.600 anuncis que es van publicar el
2025 en un portal de la comarca.
